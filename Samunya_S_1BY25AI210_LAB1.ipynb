```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
marks=np.array([65,72,81,55,90])
print("Mean:",marks.mean())
print("Maximum:",marks.max())
print("Minimum",marks.min())
print("Std.Dev.",marks.std())
```

```python
data={
    "Name": ["A", "B", "C", "D", "E"],
    "Marks": [65,72,81,55,90]
}
df = pd.DataFrame(data)
print(df)
print(df["Marks"].mean())
```

```python
plt.bar(df["Name"], df["Marks"])
plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python
plt.hist(df["Marks"])
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Distribution of Marks")
plt.show()
```

```python
result=np.random.choice(["Head","Tail"], size=10000)
heads=np.sum(result == "Head")
print("Heads:", heads)
print("Probability of Head:",heads/10000)
```

```python
import random

n = 1000
greater_than_3 = 0
even_and_greater_than_3 = 0

for i in range(n):
    die = random.randint(1, 6)

    if die > 3:
        greater_than_3 += 1

        if die % 2 == 0:
            even_and_greater_than_3 += 1

# Conditional probability
probability = even_and_greater_than_3 / greater_than_3

print("Number of rolls:", n)
print("Number greater than 3:", greater_than_3)
print("Even numbers greater than 3:", even_and_greater_than_3)
print("Experimental Conditional Probability:", probability)
```
